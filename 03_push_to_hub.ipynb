```python
from huggingface_hub import login, HfApi, create_repo
from google.colab import files
import zipfile, os
```

## Upload and unzip the adapter

Select `text-to-sql-adapter.zip` when prompted.

```python
uploaded = files.upload()
zip_name = list(uploaded.keys())[0]
with zipfile.ZipFile(zip_name, "r") as z:
    z.extractall("adapter")
print("Unzipped:", os.listdir("adapter"))
```

## Log in

Write-scoped token from [huggingface.co/settings/tokens](https://huggingface.co/settings/tokens).

```python
login(token="")
```

## Push

```python
HF_USERNAME = "Harsh3567475586"
REPO_NAME = "qwen2.5-1.5b-text-to-sql-lora"
repo_id = f"{HF_USERNAME}/{REPO_NAME}"

create_repo(repo_id, exist_ok=True)
api = HfApi()
api.upload_folder(
    folder_path="adapter",
    repo_id=repo_id,
    repo_type="model",
    commit_message="Upload QLoRA adapter for Text-to-SQL (Qwen2.5-1.5B base)",
)

print(f"Live at: https://huggingface.co/{repo_id}")
```
